```python
from collections import deque
import time

GOAL = (1, 2, 3,
        4, 5, 6,
        7, 8, 0)          # 0 represents the blank tile

MOVES = {'Up': -3, 'Down': 3, 'Left': -1, 'Right': 1}

def actions(state):
    """Return the legal moves of the blank tile."""
    i = state.index(0)
    row, col = divmod(i, 3)
    legal = []
    if row > 0: legal.append('Up')
    if row < 2: legal.append('Down')
    if col > 0: legal.append('Left')
    if col < 2: legal.append('Right')
    return legal

def result(state, action):
    """Transition model: swap the blank with the neighbouring tile."""
    i = state.index(0)
    j = i + MOVES[action]
    s = list(state)
    s[i], s[j] = s[j], s[i]
    return tuple(s)

def goal_test(state):
    return state == GOAL

def is_solvable(state):
    tiles = [t for t in state if t != 0]
    inversions = sum(1 for a in range(len(tiles))
                       for b in range(a + 1, len(tiles)) if tiles[a] > tiles[b])
    return inversions % 2 == 0

def solve_bfs(start):
    frontier = deque([(start, [])])
    explored = {start}
    expanded = 0
    while frontier:
        state, path = frontier.popleft()
        if goal_test(state):
            return path, expanded, len(explored)
        expanded += 1
        for a in actions(state):
            child = result(state, a)
            if child not in explored:
                explored.add(child)
                frontier.append((child, path + [a]))
    return None, expanded, len(explored)

def show(state):
    for r in range(3):
        print(' '.join(str(x) if x else '_' for x in state[r*3:r*3+3]))

# --- UPDATED START STATE ---
start = (8, 1, 4,
         0, 2, 7,
         6, 5, 3)

print("Start state:"); show(start)
print("Solvable:", is_solvable(start))
t0 = time.perf_counter()
path, expanded, generated = solve_bfs(start)
t1 = time.perf_counter()

print("\n--- Results ---")
print("Solution :", path)
print("Moves    :", len(path) if path else "No solution")
print("Nodes expanded :", expanded)
print("Nodes generated:", generated)
print(f"Time taken     : {(t1-t0)*1000:.2f} ms")
```

What this does:

The GOAL tuple
This is what we're trying to reach — the solved puzzle:

text
1 2 3

Notice the blank (underscore _) at the bottom-right. In code, we represent it as 0.

The board is stored as a flat tuple (a list that can't be changed), reading left-to-right, top-to-bottom:

text
Position: 0  1  2  3  4  5  6  7  8
Value:    1  2  3  4  5  6  7  8  0
The MOVES dictionary
This is the clever trick. Instead of thinking of the board as 2D, we flatten it into 1D and use position offsets:

Direction	Offset	Why
Up	−3	Going up = skipping back 3 positions (one full row)
Down	+3	Going down = skipping forward 3 positions
Left	−1	Going left = back 1 position
Right	+1	Going right = forward 1 position
Example: If the blank is at position 4 (the center), moving Up puts it at 4 − 3 = 1 (top-middle).

Why we need it: This makes moving the blank super easy — just add or subtract a number from its position.

    What this does:

This function answers: "Where can the blank legally move from here?"

Line by line:
i = state.index(0) → Find where the blank (0) currently is. It returns a number 0–8.

row, col = divmod(i, 3) → Convert that flat position into row and column.

divmod(4, 3) gives (1, 1) → position 4 is at row 1, column 1 (the center).

Row 0 = top, Row 2 = bottom

Col 0 = left, Col 2 = right

The four if checks → Add legal moves based on the blank's position:

Check	Meaning	Why
row > 0 → add Up	Blank is NOT on the top row	Can't go up if already at top
row < 2 → add Down	Blank is NOT on the bottom row	Can't go down if already at bottom
col > 0 → add Left	Blank is NOT in the leftmost column	Can't go left if already at the left
col < 2 → add Right	Blank is NOT in the rightmost column	Can't go right if already at the right
Example: Blank at position 4 (center) → row=1, col=1 → all four moves allowed.

Example: Blank at position 0 (top-left) → row=0, col=0 → only Down and Right allowed (can't go Up or Left).

Why we need it: Without this, we'd crash trying to move the blank outside the board.

  What this does:

This actually performs the move. Given a board and a direction, it swaps the blank with the neighboring tile.

Line by line:
i = state.index(0) → Where's the blank right now?

j = i + MOVES[action] → Where will the blank go?

If action is 'Up', j = i + (−3)

If action is 'Right', j = i + 1

etc.

s = list(state) → Convert the tuple to a list because tuples can't be changed. Lists can be modified.

s[i], s[j] = s[j], s[i] → Swap the two positions. This is a fancy Python one-liner for "trade the values at positions i and j."

return tuple(s) → Convert back to a tuple (immutable = safer for storing in sets/dictionaries).

Example:

Board: (1, 2, 3, 4, 0, 6, 7, 5, 8), blank at position 4

Action: 'Down' → j = 4 + 3 = 7

Swap positions 4 and 7 → (1, 2, 3, 4, 5, 6, 7, 0, 8)

Why we need it: This is the transition model — the rule that says "do this action, get this new state."

SECTION 6: is_solvable(state) — Can This Even Be Solved?


def is_solvable(state):

    tiles = [t for t in state if t != 0]
    inversions = sum(1 for a in range(len(tiles))

                       for b in range(a + 1, len(tiles)) if tiles[a] > tiles[b])

    return inversions % 2 == 0

    What this does:

This is the smart check. Not every 8-puzzle can be solved — only half of all arrangements can. This function tells you which half you're in.

What's an "inversion"?
An inversion is when a bigger tile comes before a smaller tile (ignoring the blank).

Line by line:
tiles = [t for t in state if t != 0] → Remove the blank. Keep only tiles 1–8.

The big sum(...) line → This is a nested loop in one line:

For every pair (a, b) where a is before b:

If tiles[a] > tiles[b], count it as an inversion.

Read it like: "Count every pair (a, b) where a comes earlier but a is bigger than b."

return inversions % 2 == 0 → If the count is even, return True (solvable). If odd, return False (unsolvable).

The rule:
Inversions	Solvable?
0, 2, 4, 6, 8, ... (even)	✅ Yes
1, 3, 5, 7, ... (odd)	❌ No
Example: (1, 2, 3, 4, 5, 6, 8, 7, 0) → tiles = [1,2,3,4,5,6,8,7] → one inversion (8,7) → odd → unsolvable.

Why we need it: To avoid wasting time searching an impossible state. The lab specifically says: "Make sure it never starts searching an unsolvable state."



SECTION 7: solve_bfs(start) — The Search Algorithm


def solve_bfs(start):

    frontier = deque([(start, [])])
    explored = {start}
    expanded = 0
    while frontier:

        state, path = frontier.popleft()
        if goal_test(state):
            return path, expanded, len(explored)
        expanded += 1
        for a in actions(state):
            child = result(state, a)
            if child not in explored:
                explored.add(child)
                frontier.append((child, path + [a]))
    return None, expanded, len(explored)


What this does:

This is the heart of the code. BFS = Breadth-First Search — it tries all 1-step solutions first, then all 2-step, then 3-step, etc.

Understanding BFS visually:
Think of it as exploring a tree level by level:

text
Level 0:  [Start]
Level 1:  [State A] [State B] [State C]
Level 2:  [A1] [A2] [B1] [B2] [C1] [C2]
Level 3:  ...
BFS checks all of Level 1 before Level 2. This is why it finds the shortest path first.

Line by line:
frontier = deque([(start, [])])

Frontier = the queue of states we haven't explored yet.

Each item is (state, path) where path is the list of moves to reach it.

Start with the initial state and an empty path.

explored = {start}

Explored = a set of states we've already seen.

Prevents going in circles (e.g., moving Up then Down forever).

expanded = 0

Counts how many states we've actually processed (the "work" metric).

while frontier: → Keep going while there are states to explore.

state, path = frontier.popleft()

Take the first item from the queue (FIFO!).

That's why we use deque — it's fast at popping from the left.

if goal_test(state): → Is this the goal?

Yes → return (path, expanded, len(explored)).

path = solution moves, expanded = work done, len(explored) = nodes generated.

expanded += 1 → Increment the work counter.

for a in actions(state): → Try every legal move from this state.

child = result(state, a) → Create the new state after the move.

if child not in explored: → Only add it if we've never seen it.

Why? To avoid infinite loops and duplicate work.

explored.add(child) → Mark it as seen.

frontier.append((child, path + [a]))

Add the new state to the queue.

path + [a] = append the current move a to the path list.

return None, expanded, len(explored) → If the queue empties without finding the goal, return None (no solution).

Why we need it: This is the actual problem-solving part. Everything else is setup.
