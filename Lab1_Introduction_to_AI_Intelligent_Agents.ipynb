```python
# =========================================================
# TASK 3: Random Agent vs Simple Reflex Agent Comparison
# =========================================================

import random
import statistics

# ---------------- Environment ----------------
class VacuumEnvironment:
    """Two-square world: 'A' and 'B', each Clean (0) or Dirty (1)."""

    def __init__(self):
        self.status = {'A': random.randint(0, 1), 'B': random.randint(0, 1)}
        self.agent_location = random.choice(['A', 'B'])

    def percept(self):
        return (self.agent_location, self.status[self.agent_location])

    def execute(self, action):
        if action == 'Suck':
            self.status[self.agent_location] = 0
        elif action == 'Right':
            self.agent_location = 'B'
        elif action == 'Left':
            self.agent_location = 'A'
        # NoOp does nothing

# ---------------- Reflex Agent ----------------
def reflex_vacuum_agent(percept):
    location, dirty = percept
    if dirty:
        return 'Suck'
    elif location == 'A':
        return 'Right'
    else:
        return 'Left'

# ---------------- Random Agent ----------------
def random_agent(percept):
    return random.choice(['Left', 'Right', 'Suck'])

# ---------------- Simulation Runner ----------------
def run_simulation(agent_func, steps=20, seed=None):
    if seed is not None:
        random.seed(seed)
    env = VacuumEnvironment()
    score = 0
    for _ in range(steps):
        p = env.percept()
        action = agent_func(p)
        env.execute(action)
        if action == 'Suck':
            score += 10
        elif action in ('Left', 'Right'):
            score -= 1
    return score

# ---------------- Run 10 Trials ----------------
print("=" * 60)
print("TASK 3: Random Agent vs Simple Reflex Agent (10 Trials)")
print("=" * 60)

reflex_scores = []
random_scores = []

for trial in range(1, 11):
    r_score = run_simulation(reflex_vacuum_agent, steps=20, seed=trial)
    rand_score = run_simulation(random_agent, steps=20, seed=trial)
    reflex_scores.append(r_score)
    random_scores.append(rand_score)
    print(f"Trial {trial:2d} | Reflex Score: {r_score:4d} | Random Score: {rand_score:4d}")

print("-" * 60)
print(f"Average Reflex Score : {statistics.mean(reflex_scores):.2f}")
print(f"Average Random Score : {statistics.mean(random_scores):.2f}")
print("-" * 60)

if statistics.mean(reflex_scores) > statistics.mean(random_scores):
    print("✅ Conclusion: Reflex agent outperforms the random agent.")
else:
    print("⚠️ Conclusion: Random agent performed better (unexpected).")
```

```python
# =========================================================
# TASK 4: Model-Based Reflex Agent
# Remembers cleaned locations & stops (NoOp) when done
# =========================================================

import random

class VacuumEnvironment:
    def __init__(self):
        self.status = {'A': random.randint(0, 1), 'B': random.randint(0, 1)}
        self.agent_location = random.choice(['A', 'B'])

    def percept(self):
        return (self.agent_location, self.status[self.agent_location])

    def execute(self, action):
        if action == 'Suck':
            self.status[self.agent_location] = 0
        elif action == 'Right':
            self.agent_location = 'B'
        elif action == 'Left':
            self.agent_location = 'A'
        # NoOp → no change

# ---------------- Model-Based Agent ----------------
class ModelBasedVacuumAgent:
    """Keeps internal state: which squares have been cleaned."""

    def __init__(self):
        self.model = {'A': 'Unknown', 'B': 'Unknown'}

    def act(self, percept):
        location, dirty = percept

        # Update internal model
        self.model[location] = 'Dirty' if dirty else 'Clean'

        # Decide action
        if dirty:
            return 'Suck'
        # If both locations known clean → stop
        if self.model['A'] == 'Clean' and self.model['B'] == 'Clean':
            return 'NoOp'
        # Otherwise move to the other square
        return 'Right' if location == 'A' else 'Left'

# ---------------- Simulation ----------------
def run_model_based_agent(steps=20, seed=None):
    if seed is not None:
        random.seed(seed)
    env = VacuumEnvironment()
    agent = ModelBasedVacuumAgent()
    score = 0
    history = []

    for step in range(1, steps + 1):
        p = env.percept()
        action = agent.act(p)
        env.execute(action)

        if action == 'Suck':
            score += 10
        elif action in ('Left', 'Right'):
            score -= 1
        # NoOp → no penalty

        history.append((step, p, action, dict(env.status), score))

        if action == 'NoOp':
            break   # Agent decided to stop

    return score, history

# ---------------- Comparison ----------------
print("=" * 70)
print("TASK 4: Model-Based Agent vs Simple Reflex Agent")
print("=" * 70)

def reflex_vacuum_agent(percept):
    location, dirty = percept
    if dirty:
        return 'Suck'
    elif location == 'A':
        return 'Right'
    else:
        return 'Left'

def run_reflex(seed=None, steps=20):
    if seed is not None:
        random.seed(seed)
    env = VacuumEnvironment()
    score = 0
    for _ in range(steps):
        p = env.percept()
        action = reflex_vacuum_agent(p)
        env.execute(action)
        if action == 'Suck':
            score += 10
        elif action in ('Left', 'Right'):
            score -= 1
    return score

model_scores, reflex_scores = [], []

for trial in range(1, 11):
    m_score, _ = run_model_based_agent(seed=trial)
    r_score = run_reflex(seed=trial)
    model_scores.append(m_score)
    reflex_scores.append(r_score)
    print(f"Trial {trial:2d} | Model-Based: {m_score:4d} | Reflex: {r_score:4d}")

print("-" * 70)
print(f"Average Model-Based Score : {sum(model_scores)/len(model_scores):.2f}")
print(f"Average Reflex Score      : {sum(reflex_scores)/len(reflex_scores):.2f}")

# ---------------- Detailed trace for one run ----------------
print("\n" + "=" * 70)
print("DETAILED TRACE (Model-Based Agent, Seed = 1)")
print("=" * 70)
score, history = run_model_based_agent(seed=1)
for step, percept, action, world, sc in history:
    print(f"Step {step}: percept={percept} -> action={action:5s} | world={world} | score={sc}")
print(f"\nFinal Score: {score}")
```

```python
# =========================================================
# POST-LAB TASK: 3x3 Grid Model-Based Reflex Agent
# Visits every room, cleans all dirt, reports metrics
# =========================================================

import random

class GridEnvironment:
    """3x3 grid. Each cell is Clean (0) or Dirty (1)."""

    def __init__(self):
        self.grid = [[random.randint(0, 1) for _ in range(3)] for _ in range(3)]
        self.row = random.randint(0, 2)
        self.col = random.randint(0, 2)

    def percept(self):
        return (self.row, self.col, self.grid[self.row][self.col])

    def execute(self, action):
        if action == 'Suck':
            self.grid[self.row][self.col] = 0
        elif action == 'Up' and self.row > 0:
            self.row -= 1
        elif action == 'Down' and self.row < 2:
            self.row += 1
        elif action == 'Left' and self.col > 0:
            self.col -= 1
        elif action == 'Right' and self.col < 2:
            self.col += 1

    def all_clean(self):
        return all(cell == 0 for row in self.grid for cell in row)

    def dirty_count(self):
        return sum(cell for row in self.grid for cell in row)


class GridModelBasedAgent:
    """Systematically visits every cell and cleans it."""

    def __init__(self):
        self.visited = set()
        self.path = []  # cells to visit in order

    def act(self, percept):
        row, col, dirty = percept

        # If current cell is dirty, clean it first
        if dirty:
            return 'Suck'

        # Mark current cell as visited
        self.visited.add((row, col))

        # Decide next move: systematically sweep the grid
        if col < 2:
            return 'Right'
        elif col == 2 and row < 2:
            return 'Down'
        elif row == 2 and col == 2:
            return 'NoOp'
        return 'NoOp'


def run_grid_simulation(seed=None, max_steps=100):
    if seed is not None:
        random.seed(seed)

    env = GridEnvironment()
    agent = GridModelBasedAgent()

    moves = 0
    cleaned = 0
    score = 0

    print("Initial Grid:")
    for row in env.grid:
        print(row)
    print(f"Agent starts at: ({env.row}, {env.col})")
    print("-" * 50)

    for step in range(1, max_steps + 1):
        p = env.percept()
        action = agent.act(p)

        before = env.dirty_count()
        env.execute(action)
        after = env.dirty_count()

        if action == 'Suck':
            if after < before:
                cleaned += 1
                score += 10
        elif action in ('Up', 'Down', 'Left', 'Right'):
            moves += 1
            score -= 1

        print(f"Step {step:2d}: percept={p} -> action={action:5s} | grid={env.grid} | score={score}")

        if env.all_clean() and action == 'NoOp':
            print("✅ All rooms cleaned. Agent stopped.")
            break

    print("-" * 50)
    print(f"Total Moves        : {moves}")
    print(f"Rooms Cleaned      : {cleaned}")
    print(f"Performance Score  : {score}")
    return moves, cleaned, score


# Run the simulation
print("=" * 60)
print("POST-LAB TASK: 3x3 Grid Model-Based Agent")
print("=" * 60)
run_grid_simulation(seed=42)
```

# Viva Questions — Answers

## Q1. What is the difference between an agent function and an agent program?

- **Agent Function:** A mathematical mapping from percept sequences to actions. It describes *what* the agent should do.
- **Agent Program:** The concrete implementation (in code) of the agent function running on a physical architecture. It describes *how* the agent does it.

**Example:** In the vacuum world, the agent function maps `[B, Dirty] → Suck`. The agent program is the Python `if-else` code that implements this rule.

---

## Q2. Why is a simple reflex agent not suitable for a partially observable environment?

A simple reflex agent only responds to the **current percept** and has no memory of the past. In a partially observable environment, the current percept does not reveal the full state of the world — so the agent may keep repeating useless actions (e.g., moving Right even after both squares are clean) because it cannot "remember" what it has already done.

---

## Q3. Is the vacuum world in this lab deterministic or stochastic? Justify.

The **agent's actions** (Suck, Left, Right) are **deterministic** — they always produce the same result.

However, the **environment's initial state** is **stochastic** because dirt is placed randomly using `random.randint(0, 1)`. So the world can be considered a **stochastic environment with deterministic action outcomes**.

---

## Q4. Give an example of a utility-based agent from daily life.

**A self-driving car** is a utility-based agent. It doesn't just try to reach the destination (goal-based) — it maximizes a **utility function** that balances safety, speed, fuel efficiency, passenger comfort, and traffic laws.

Another example: **A Netflix recommendation system** maximizes user satisfaction (utility) based on watch history, ratings, and time of day.
