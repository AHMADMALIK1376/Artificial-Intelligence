```python
# =========================================================
# TASK 3 (IMPROVED): Random vs Reflex Agent
# Stops when the world becomes clean (fair comparison)
# =========================================================

import random
import statistics

class VacuumEnvironment:
    def __init__(self):
        self.status = {'A': random.randint(0, 1), 'B': random.randint(0, 1)}
        self.agent_location = random.choice(['A', 'B'])

    def percept(self):
        return (self.agent_location, self.status[self.agent_location])

    def execute(self, action):
        if action == 'Suck':
            self.status[self.agent_location] = 0
        elif action == 'Right':
            self.agent_location = 'B'
        elif action == 'Left':
            self.agent_location = 'A'

    def is_clean(self):
        return self.status['A'] == 0 and self.status['B'] == 0


def reflex_vacuum_agent(percept):
    location, dirty = percept
    if dirty:
        return 'Suck'
    elif location == 'A':
        return 'Right'
    else:
        return 'Left'


def random_agent(percept):
    return random.choice(['Left', 'Right', 'Suck'])


def run_simulation(agent_func, max_steps=50, seed=None):
    if seed is not None:
        random.seed(seed)
    env = VacuumEnvironment()
    score = 0
    steps_taken = 0

    for _ in range(max_steps):
        if env.is_clean():
            break
        p = env.percept()
        action = agent_func(p)
        env.execute(action)
        steps_taken += 1
        if action == 'Suck':
            score += 10
        elif action in ('Left', 'Right'):
            score -= 1

    return score, steps_taken


# ---------------- Run 10 Trials ----------------
print("=" * 70)
print("TASK 3 (IMPROVED): Random vs Reflex Agent (10 Trials)")
print("Stops when world becomes CLEAN — fair comparison")
print("=" * 70)

reflex_scores, random_scores = [], []
reflex_steps, random_steps = [], []

for trial in range(1, 11):
    r_score, r_steps = run_simulation(reflex_vacuum_agent, seed=trial)
    rand_score, rand_steps = run_simulation(random_agent, seed=trial)
    reflex_scores.append(r_score)
    random_scores.append(rand_score)
    reflex_steps.append(r_steps)
    random_steps.append(rand_steps)
    print(f"Trial {trial:2d} | Reflex: {r_score:4d} ({r_steps:2d} steps) | Random: {rand_score:4d} ({rand_steps:2d} steps)")

print("-" * 70)
print(f"Average Reflex Score : {statistics.mean(reflex_scores):.2f}")
print(f"Average Random Score : {statistics.mean(random_scores):.2f}")
print(f"Average Reflex Steps : {statistics.mean(reflex_steps):.2f}")
print(f"Average Random Steps : {statistics.mean(random_steps):.2f}")
print("-" * 70)

if statistics.mean(reflex_scores) > statistics.mean(random_scores):
    print("✅ Conclusion: Reflex agent outperforms the random agent.")
else:
    print("⚠️ Conclusion: Random agent performed better.")
```

```python
# =========================================================
# TASK 4: Model-Based Reflex Agent
# Remembers cleaned locations & stops when done
# =========================================================

import random

class VacuumEnvironment:
    def __init__(self):
        self.status = {'A': random.randint(0, 1), 'B': random.randint(0, 1)}
        self.agent_location = random.choice(['A', 'B'])

    def percept(self):
        return (self.agent_location, self.status[self.agent_location])

    def execute(self, action):
        if action == 'Suck':
            self.status[self.agent_location] = 0
        elif action == 'Right':
            self.agent_location = 'B'
        elif action == 'Left':
            self.agent_location = 'A'


class ModelBasedVacuumAgent:
    """Keeps internal state: which squares have been cleaned."""

    def __init__(self):
        self.model = {'A': 'Unknown', 'B': 'Unknown'}

    def act(self, percept):
        location, dirty = percept
        self.model[location] = 'Dirty' if dirty else 'Clean'

        if dirty:
            return 'Suck'
        if self.model['A'] == 'Clean' and self.model['B'] == 'Clean':
            return 'NoOp'
        return 'Right' if location == 'A' else 'Left'


def run_model_based_agent(steps=20, seed=None):
    if seed is not None:
        random.seed(seed)
    env = VacuumEnvironment()
    agent = ModelBasedVacuumAgent()
    score = 0
    history = []

    for step in range(1, steps + 1):
        p = env.percept()
        action = agent.act(p)
        env.execute(action)

        if action == 'Suck':
            score += 10
        elif action in ('Left', 'Right'):
            score -= 1

        history.append((step, p, action, dict(env.status), score))

        if action == 'NoOp':
            break

    return score, history


def reflex_vacuum_agent(percept):
    location, dirty = percept
    if dirty:
        return 'Suck'
    elif location == 'A':
        return 'Right'
    else:
        return 'Left'


def run_reflex(seed=None, steps=20):
    if seed is not None:
        random.seed(seed)
    env = VacuumEnvironment()
    score = 0
    for _ in range(steps):
        p = env.percept()
        action = reflex_vacuum_agent(p)
        env.execute(action)
        if action == 'Suck':
            score += 10
        elif action in ('Left', 'Right'):
            score -= 1
    return score


# ---------------- Comparison ----------------
print("=" * 70)
print("TASK 4: Model-Based Agent vs Simple Reflex Agent")
print("=" * 70)

model_scores, reflex_scores = [], []

for trial in range(1, 11):
    m_score, _ = run_model_based_agent(seed=trial)
    r_score = run_reflex(seed=trial)
    model_scores.append(m_score)
    reflex_scores.append(r_score)
    print(f"Trial {trial:2d} | Model-Based: {m_score:4d} | Reflex: {r_score:4d}")

print("-" * 70)
print(f"Average Model-Based Score : {sum(model_scores)/len(model_scores):.2f}")
print(f"Average Reflex Score      : {sum(reflex_scores)/len(reflex_scores):.2f}")

# ---------------- Detailed trace ----------------
print("\n" + "=" * 70)
print("DETAILED TRACE (Model-Based Agent, Seed = 1)")
print("=" * 70)
score, history = run_model_based_agent(seed=1)
for step, percept, action, world, sc in history:
    print(f"Step {step}: percept={percept} -> action={action:5s} | world={world} | score={sc}")
print(f"\nFinal Score: {score}")
```

```python
# =========================================================
# POST-LAB TASK: 3x3 Grid Model-Based Reflex Agent
# =========================================================

import random

class GridEnvironment:
    def __init__(self):
        self.grid = [[random.randint(0, 1) for _ in range(3)] for _ in range(3)]
        self.row = random.randint(0, 2)
        self.col = random.randint(0, 2)

    def percept(self):
        return (self.row, self.col, self.grid[self.row][self.col])

    def execute(self, action):
        if action == 'Suck':
            self.grid[self.row][self.col] = 0
        elif action == 'Up' and self.row > 0:
            self.row -= 1
        elif action == 'Down' and self.row < 2:
            self.row += 1
        elif action == 'Left' and self.col > 0:
            self.col -= 1
        elif action == 'Right' and self.col < 2:
            self.col += 1

    def all_clean(self):
        return all(cell == 0 for row in self.grid for cell in row)

    def dirty_count(self):
        return sum(cell for row in self.grid for cell in row)


class GridModelBasedAgent:
    def __init__(self):
        self.visited = set()

    def act(self, percept):
        row, col, dirty = percept
        if dirty:
            return 'Suck'
        self.visited.add((row, col))

        if col < 2:
            return 'Right'
        elif col == 2 and row < 2:
            return 'Down'
        else:
            return 'NoOp'


def run_grid_simulation(seed=None, max_steps=100):
    if seed is not None:
        random.seed(seed)

    env = GridEnvironment()
    agent = GridModelBasedAgent()

    moves = 0
    cleaned = 0
    score = 0

    print("Initial Grid:")
    for row in env.grid:
        print(row)
    print(f"Agent starts at: ({env.row}, {env.col})")
    print("-" * 50)

    for step in range(1, max_steps + 1):
        p = env.percept()
        action = agent.act(p)

        before = env.dirty_count()
        env.execute(action)
        after = env.dirty_count()

        if action == 'Suck':
            if after < before:
                cleaned += 1
                score += 10
        elif action in ('Up', 'Down', 'Left', 'Right'):
            moves += 1
            score -= 1

        print(f"Step {step:2d}: percept={p} -> action={action:5s} | grid={env.grid} | score={score}")

        if env.all_clean() and action == 'NoOp':
            print("✅ All rooms cleaned. Agent stopped.")
            break

    print("-" * 50)
    print(f"Total Moves        : {moves}")
    print(f"Rooms Cleaned      : {cleaned}")
    print(f"Performance Score  : {score}")
    return moves, cleaned, score


print("=" * 60)
print("POST-LAB TASK: 3x3 Grid Model-Based Agent")
print("=" * 60)
run_grid_simulation(seed=42)
```

# Viva Questions — Answers

## Q1. Agent Function vs Agent Program
- **Agent Function:** Mathematical mapping from percept sequences to actions — *what* the agent should do.
- **Agent Program:** Concrete implementation of the agent function running on a physical architecture — *how* the agent does it.

**Example:** Vacuum world maps `[B, Dirty] → Suck` (function). The Python `if-else` code implements it (program).

---

## Q2. Why is a simple reflex agent not suitable for partially observable environments?
A simple reflex agent only responds to the **current percept** and has **no memory**. In a partially observable environment, the current percept does not reveal the full world state — so the agent may repeat useless actions because it cannot "remember" what it has done.

---

## Q3. Is the vacuum world deterministic or stochastic?
- **Agent actions** are deterministic (Suck always cleans; Left/Right always move predictably).
- **Initial state** is stochastic (dirt is placed randomly via `random.randint`).
- Therefore: **a stochastic environment with deterministic action outcomes**.

---

## Q4. Example of a utility-based agent from daily life
**Self-driving car:** Maximizes a utility function balancing safety, speed, fuel efficiency, comfort, and traffic laws.

**Another example:** Netflix recommendation system — maximizes user satisfaction based on watch history.
