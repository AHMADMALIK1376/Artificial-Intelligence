```python
# =========================================================
# COMPREHENSIVE PYTHON BASICS ASSIGNMENT
# Subject: Artificial Intelligence and Technology
# Covers: Variables, Input, If-Else, Switch (Match), Loops,
#         1D Arrays, 2D Arrays, Functions, Break/Continue
# =========================================================

import time

# --- 1. FUNCTIONS (Reusable blocks of code) ---
def print_header(title):
    """Prints a beautiful header"""
    print("\n" + "="*50)
    print(f"  {title.upper()}")
    print("="*50)

def get_user_name():
    """Asks for user input with validation"""
    while True:
        name = input("Enter your name (or type 'exit' to quit): ").strip()
        if name.lower() == 'exit':
            return None
        if len(name) > 0:
            return name
        print("❌ Name cannot be empty. Please try again.")

# --- MAIN PROGRAM STARTS HERE ---
print_header("Welcome to the AI Fundamentals Checker")

# --- 2. VARIABLES & USER INPUT ---
user_name = get_user_name()
if user_name is None:
    print("Goodbye!")
    exit()

# Asking for age with error handling (Try/Except)
while True:
    try:
        age = int(input(f"Hello {user_name}, please enter your age: "))
        if age < 0:
            print("Age cannot be negative.")
            continue
        break
    except ValueError:
        print("❌ Invalid input! Please enter a number.")

# --- 3. IF-ELSE STATEMENTS (Conditionals) ---
print_header("Age Verification")
if age < 13:
    category = "Child"
elif age >= 13 and age < 20:
    category = "Teenager"
elif age >= 20 and age < 60:
    category = "Adult"
else:
    category = "Senior Citizen"

print(f"Based on your age ({age}), you are categorized as: **{category}**")

# --- 4. SWITCH STATEMENT (Python Equivalent) ---
# Python 3.10+ uses "match-case" (Structural Pattern Matching)
print_header("Switch Statement Equivalent (Match-Case)")
print("Select a field of AI to explore:")
print("1. Machine Learning")
print("2. Neural Networks")
print("3. Natural Language Processing")

choice = input("Enter your choice (1-3): ")

match choice:
    case "1":
        selected_field = "Machine Learning"
        description = "Algorithms that learn from data."
    case "2":
        selected_field = "Neural Networks"
        description = "Brain-inspired systems for deep learning."
    case "3":
        selected_field = "Natural Language Processing"
        description = "Teaching computers to understand human language."
    case _:
        selected_field = "Unknown"
        description = "Invalid choice. Please select 1, 2, or 3 next time."

print(f"✅ You selected: {selected_field}")
print(f"📖 Description: {description}")

# --- 5. ARRAYS (Lists) & LOOPS (For, While, Break, Continue) ---
print_header("Array (List) Operations")

# 1D Array (List)
ai_tools = ["TensorFlow", "PyTorch", "Scikit-Learn", "Keras", "Pandas"]

print("Current AI Tools in our array:")
for tool in ai_tools:
    print(f" - {tool}")

# Adding user input to the array
new_tool = input("\nEnter a new AI tool to add to the array: ")
ai_tools.append(new_tool)

# Demonstrating Break and Continue
print("\nLooping through updated array (skipping 'Keras', stopping at 'Pandas'):")
for tool in ai_tools:
    if tool == "Keras":
        print("   (Skipping Keras...)")
        continue  # Skip this iteration
    if tool == "Pandas":
        print("   (Found Pandas! Stopping loop...)")
        break     # Exit the loop entirely
    print(f"   Processing: {tool}")

# --- 6. 2D ARRAYS (Matrices) & NESTED LOOPS ---
print_header("2D Array (Matrix) - Neural Network Weights")

# Creating a 2D Array (3x3 Matrix)
# Imagine this represents weights in a neural network layer
weights_matrix = [
    [0.5, 0.2, 0.8],
    [0.1, 0.9, 0.4],
    [0.7, 0.3, 0.6]
]

print("Original 2D Matrix:")
for row in weights_matrix:
    print(row)

# Modifying a value with user input (Nested Loops + Input)
print("\nLet's update a specific weight in the matrix.")
while True:
    try:
        row_idx = int(input("Enter row index (0-2): "))
        col_idx = int(input("Enter column index (0-2): "))
        if 0 <= row_idx <= 2 and 0 <= col_idx <= 2:
            break
        else:
            print("❌ Index out of bounds. Try again.")
    except ValueError:
        print("❌ Please enter valid integers.")

new_value = float(input(f"Enter new weight value for position [{row_idx}][{col_idx}]: "))
weights_matrix[row_idx][col_idx] = new_value

print("\nUpdated 2D Matrix:")
for row in weights_matrix:
    # We use a nested loop to format the output beautifully
    formatted_row = " | ".join([f"{val:.2f}" for val in row])
    print(f"[ {formatted_row} ]")

# --- 7. WHILE LOOP WITH CONDITION (Final Loop Demo) ---
print_header("Final Countdown (While Loop)")
counter = 3
while counter > 0:
    print(f"T-minus {counter}...")
    time.sleep(0.5) # Adds a small delay for effect
    counter -= 1
print("🚀 System Ready for AI Deployment!")

print("\n" + "="*50)
print("ASSIGNMENT COMPLETE. THANK YOU!")
print("="*50)
```
